## 确认环境

运行下面的单元格，确认你确实在 Linux 里。第一行应当显示 `Linux`。

````python
%%bash
uname -sm
grep PRETTY_NAME /etc/os-release
echo "当前用户: $(whoami)"
echo "默认 Shell: $SHELL"
````

## 怎样使用这些 notebook

**1. 演示单元格是安全的。** 它们都在临时目录 `/tmp/linux-demo` 里运行，不会影响你的文件。

**2. 每个 `%%bash` 单元格都是一个全新的 Shell。** 上一个单元格里的 `cd` 和变量，不会带到下一个单元格，所以你会看到每个单元格开头都有一句 `cd`。真实终端里不需要这样，`cd` 之后会一直留在那个目录。

````python
%%bash
cd /tmp
pwd
````

````python
%%bash
pwd
````

第二个单元格的输出不是 `/tmp`，这就是上面说的情况。

**3. 不要只运行单元格。** 看懂演示后，把命令在真实终端里亲手敲一遍，换个参数试试。终端靠手熟，只看是学不会的。

**4. 交互式命令要到终端里运行。** `less`、`nano`、`top`、`ssh`、`tmux` 这类命令需要和你互动，在单元格里运行会卡住，它们只写在说明文字里。如果某个单元格一直不结束，点「中断」即可。
