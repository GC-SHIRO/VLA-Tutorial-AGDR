## 为什么要学终端

机器人和深度学习的工作几乎都离不开 Linux 终端：

- 实验室的训练服务器通常没有图形界面，只能通过终端操作
- ROS2 和 LeRobot 的主要运行环境都是 Linux，日常要靠命令启动程序
- 命令可以写成脚本反复运行，比点鼠标更容易重复和自动化

## 终端、Shell、命令

这三个词经常被混用，其实是不同的东西：

| 名称 | 是什么 | 例子 |
| --- | --- | --- |
| 终端（Terminal） | 显示文字、接收键盘输入的窗口 | GNOME Terminal、VS Code 内置终端 |
| Shell | 读懂你输入的命令并执行的程序 | bash、zsh |
| 命令（Command） | 被 Shell 执行的具体程序或内建功能 | `ls`、`cd` |

Ubuntu 默认的 Shell 是 **bash**，本教程也以它为准。

## 读懂提示符

打开终端后，你会看到类似这样的一行：

```text
shiro@ubuntu:~/projects$
```

| 部分 | 含义 |
| --- | --- |
| `shiro` | 当前用户名 |
| `ubuntu` | 主机名，连接多台服务器时用它分辨自己在哪 |
| `~/projects` | 当前所在目录 |
| `$` | 普通用户；如果是 `#`，说明你是 root，要格外小心 |

## 命令的结构

```text
ls   -l -a   /etc
命令  选项     参数
```

- **命令**：要运行什么
- **选项**：怎么运行，通常以 `-` 或 `--` 开头。短选项可以合并，`-l -a` 等于 `-la`
- **参数**：对什么运行

Linux **区分大小写**：`file.txt` 和 `File.txt` 是两个文件，`-a` 和 `-A` 也是两个选项。

````python
%%bash
ls -la /etc | head -n 5
````

上面的 `| head -n 5` 表示只看前 5 行，第 04 节会讲。

## 忘了怎么用？先查文档

| 方法 | 例子 | 说明 |
| --- | --- | --- |
| `--help` | `ls --help` | 简短的用法说明，大多数命令都有 |
| `man` | `man ls` | 完整手册，在终端里运行，按 `q` 退出 |
| `type` | `type cd` | 查看它是外部程序还是 Shell 内建命令 |

遇到陌生命令，先看 `--help`，再问 Agent。

````python
%%bash
ls --help | head -n 8
type cd
type ls
````

## 文件系统的结构

Linux 只有一棵目录树，从根目录 `/` 开始，没有 C 盘、D 盘的概念。

| 目录 | 放什么 |
| --- | --- |
| `/` | 根目录，一切的起点 |
| `/home/用户名` | 你的主目录，个人文件都放这里 |
| `/etc` | 系统配置文件 |
| `/usr` | 安装的程序和库 |
| `/var` | 经常变化的数据，例如日志 `/var/log` |
| `/tmp` | 临时文件，重启后通常会被清空 |
| `/dev` | 设备文件；机器人常用的串口、摄像头就在这里，例如 `/dev/ttyUSB0`、`/dev/video0` |

````python
%%bash
ls /
````

## 路径：绝对路径与相对路径

| 写法 | 含义 |
| --- | --- |
| `/home/shiro/a.txt` | **绝对路径**，以 `/` 开头，从根目录算起，在哪都能用 |
| `projects/a.txt` | **相对路径**，从当前目录算起 |
| `.` | 当前目录 |
| `..` | 上一级目录 |
| `~` | 你的主目录 |

分隔符是 `/`，不是 Windows 的 `\`。

## pwd、cd、ls

| 命令 | 作用 |
| --- | --- |
| `pwd` | 显示当前目录 |
| `cd <目录>` | 进入目录 |
| `cd` 或 `cd ~` | 回到主目录 |
| `cd ..` | 上一级 |
| `cd -` | 回到上一次所在的目录 |
| `ls` | 列出当前目录内容 |
| `ls -l` | 显示详情：权限、大小、修改时间 |
| `ls -a` | 显示隐藏文件（文件名以 `.` 开头） |
| `ls -lh` | 大小以 K、M、G 显示 |

````python
%%bash
rm -rf /tmp/linux-demo && mkdir -p /tmp/linux-demo/projects/robot && cd /tmp/linux-demo
touch notes.txt .secret projects/robot/arm.py
pwd
cd projects/robot
pwd
cd ../..
pwd
````

`cd ../..` 向上走了两级，回到了 `/tmp/linux-demo`。再看看隐藏文件：

````python
%%bash
cd /tmp/linux-demo
ls
echo ---
ls -a
echo ---
ls -lh projects/robot
````

`ls` 看不到 `.secret`，`ls -a` 才能看到。很多配置文件都是隐藏文件，比如第 06 节的 `~/.bashrc`。
