## 常用命令

| 命令 | 作用 |
| --- | --- |
| `mkdir <目录>` | 创建目录 |
| `mkdir -p a/b/c` | 连同不存在的上级目录一起创建 |
| `touch <文件>` | 创建空文件；文件已存在时只更新修改时间 |
| `cp <源> <目标>` | 复制文件 |
| `cp -r <源> <目标>` | 复制目录 |
| `mv <源> <目标>` | 移动或重命名 |
| `rm <文件>` | 删除文件 |
| `rm -r <目录>` | 删除目录及其全部内容 |

````python
%%bash
rm -rf /tmp/linux-demo && mkdir -p /tmp/linux-demo && cd /tmp/linux-demo
mkdir -p data/raw data/processed
touch data/raw/ep1.txt data/raw/ep2.txt
ls -R
````

`ls -R` 递归列出子目录。下面试试复制和重命名：

````python
%%bash
cd /tmp/linux-demo
cp data/raw/ep1.txt data/processed/
mv data/raw/ep2.txt data/raw/episode2.txt
cp -r data backup
ls -R
````

`mv` 既能移动也能重命名，原理相同：把文件放到新的路径。

## 删除没有回收站

`rm` 不经过回收站，**删了基本就找不回来了**。覆盖也一样：`cp` 和 `mv` 遇到同名文件，默认直接覆盖，不会询问。

几个保护自己的做法：

| 写法 | 作用 |
| --- | --- |
| `rm -i <文件>` | 删除前逐个确认 |
| `cp -i` / `mv -i` | 要覆盖时先确认 |
| `mv -n` | 目标已存在时不覆盖 |
| 先 `ls <同样的路径>` | 看清会匹配到什么，再把 `ls` 换成 `rm` |

下面的演示中，`mv` 默默覆盖了已有文件：

````python
%%bash
cd /tmp/linux-demo
echo "重要内容" > a.txt
echo "其他内容" > b.txt
mv b.txt a.txt
cat a.txt
ls
````

`a.txt` 原来的内容已经没了，也没有任何提示。

> ⚠️ `rm -rf` 里的 `-f` 表示强制、不询问。`rm -rf /` 或 `rm -rf ~` 这类命令会造成灾难性后果。路径里用了变量或通配符时，尤其要先确认。第 10 节有完整的安全清单。

## 通配符

**通配符由 Shell 展开**：命令运行之前，Shell 就把它替换成了匹配到的文件名。

| 通配符 | 匹配 | 例子 |
| --- | --- | --- |
| `*` | 任意长度的任意字符 | `*.py`：所有 py 文件 |
| `?` | 恰好一个字符 | `ep?.txt`：`ep1.txt`，但不匹配 `ep10.txt` |
| `[abc]` | 括号内任一字符 | `ep[12].txt` |
| `[0-9]` | 范围内任一字符 | `ep[0-9].txt` |

`*` 默认不匹配以 `.` 开头的隐藏文件。

````python
%%bash
rm -rf /tmp/linux-demo && mkdir -p /tmp/linux-demo && cd /tmp/linux-demo
touch ep1.txt ep2.txt ep10.txt model.pt notes.md
echo *.txt
echo ep?.txt
echo ep[12].txt
````

上面用 `echo` 就能看到展开结果。这也是一个好习惯：用 `echo` 或 `ls` 预览通配符，确认后再交给 `rm`。

## 带空格的路径和引号

Shell 用空格分隔参数，所以路径里有空格时必须处理：

````python
%%bash
cd /tmp/linux-demo
mkdir "my data"
ls my data 2>&1 || true
echo ---
ls "my data"
echo "(上面没有输出，因为目录是空的，但这次没有报错)"
````

不加引号时，Shell 把它当成两个参数 `my` 和 `data`，所以报错。三种写法等价：

```bash
cd "2. Linux终端操作"     # 双引号（推荐）
cd '2. Linux终端操作'     # 单引号
cd 2.\ Linux终端操作      # 反斜杠转义
```

**单引号和双引号的区别**：双引号里的变量会展开，单引号里的所有内容都原样保留。

````python
%%bash
name=shiro
echo "你好 $name"
echo '你好 $name'
````

> 自己命名文件时，建议只用英文字母、数字、`_`、`-`，不用空格，可以省去很多麻烦。
