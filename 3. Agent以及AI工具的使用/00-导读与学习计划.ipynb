## 配置 API

本章的代码通过 **环境变量** 读取 API 配置，key 不会写进任何文件。共有 6 个变量，分别对应 01 节会讲到的两种接口格式：

| 变量 | 含义 | 示例 |
| --- | --- | --- |
| `OPENAI_BASE_URL` | OpenAI 格式接口的地址 | `https://api.openai.com/v1` |
| `OPENAI_API_KEY` | 该接口的 key | `sk-...` |
| `OPENAI_MODEL` | 使用的模型名 | 以服务商文档为准 |
| `ANTHROPIC_BASE_URL` | Anthropic 格式接口的地址 | `https://api.anthropic.com` |
| `ANTHROPIC_API_KEY` | 该接口的 key | `sk-ant-...` |
| `ANTHROPIC_MODEL` | 使用的模型名 | 以服务商文档为准 |

不少国内外服务商同时兼容这两种格式，这时两组变量可以填同一个 key，只是地址不同。**地址和模型名以服务商的官方文档为准**，不要照抄网上的旧教程。

如果需要使用模型，可以使用实验室的API，请找 **Shiro** 获取 

**设置方法**：

macOS / Linux：把下面几行加到 `~/.bashrc`（zsh 用户是 `~/.zshrc`）末尾，然后运行 `source ~/.bashrc`。第 2 章第 06 节讲过这个文件。

```bash
export OPENAI_BASE_URL="https://..."
export OPENAI_API_KEY="你的 key"
export OPENAI_MODEL="模型名"
```

Windows PowerShell：下面的写法会永久保存到当前用户，设置后要**重新打开** PowerShell 才生效。

```powershell
[Environment]::SetEnvironmentVariable("OPENAI_BASE_URL", "https://...", "User")
[Environment]::SetEnvironmentVariable("OPENAI_API_KEY", "你的 key", "User")
[Environment]::SetEnvironmentVariable("OPENAI_MODEL", "模型名", "User")
```

Anthropic 的三个变量同理。设置完后**完全退出并重新打开 VS Code / Jupyter**，notebook 才能读到新的环境变量。

> ⚠️ **API key 等同于密码，而且直接关联付费额度。**
>
> - 不要写进 notebook、代码文件或截图里，更不要提交到 Git。推送到 GitHub 的 key 几分钟内就可能被扫描盗用
> - 不要发到群聊里，也不要粘贴给 Agent
> - 一旦泄露，立刻到服务商的控制台作废这个 key，再生成新的

## 确认环境

运行下面的单元格。它只显示 key 的前后几位，不会打印完整的 key。

````python
import os
import sys

print("Python 版本:", sys.version.split()[0])
if sys.version_info < (3, 10):
    print("  ⚠️ 建议升级到 3.10 或更高版本")


def mask(value):
    if not value:
        return "(未设置)"
    if len(value) <= 8:
        return "(已设置)"
    return value[:4] + "..." + value[-4:]


for name in ["OPENAI_BASE_URL", "OPENAI_MODEL", "ANTHROPIC_BASE_URL", "ANTHROPIC_MODEL"]:
    print(f"{name:20s}", os.environ.get(name, "(未设置)"))
for name in ["OPENAI_API_KEY", "ANTHROPIC_API_KEY"]:
    print(f"{name:20s}", mask(os.environ.get(name, "")))
````
